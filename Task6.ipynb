```python
#1.
def smart_parking_system(capacity, logs):
    current_vehicles = 0
    peak_vehicles = 0

    for log in logs:
        if log == "IN":
            current_vehicles += 1
        elif log == "OUT":
            if current_vehicles > 0:
                current_vehicles -= 1

        # Track peak usage
        if current_vehicles > peak_vehicles:
            peak_vehicles = current_vehicles

    print("Currently Parked Vehicles:", current_vehicles)
    print("Peak Parking Usage:", peak_vehicles)

    if current_vehicles > capacity:
        print("Parking Status: Full - Capacity Exceeded!")
    else:
        print("Parking Status: Available")

capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking_system(capacity, logs)
```

```python
#2.
def estimate_delivery_time(distance, traffic, weather):

    # Base time calculation
    base_time = distance * 5

    # Traffic adjustment
    if traffic == "Low":
        base_time += 5
    elif traffic == "Medium":
        base_time += 10
    elif traffic == "High":
        base_time += 15

    # Weather adjustment
    if weather == "Rainy":
        base_time += 10
    elif weather == "Storm":
        base_time += 20

    print("Estimated Delivery Time:", base_time, "minutes")
# Sample Input
distance = 8
traffic = "High"
weather = "Rainy"

estimate_delivery_time(distance, traffic, weather)
```

```python
#3.
def seat_occupancy_analyzer(total_seats, booked_seats):

    booked_count = len(booked_seats)

    occupancy_percentage = (booked_count / total_seats) * 100

    print("Occupancy:", int(occupancy_percentage), "%")

    # Show Status
    if occupancy_percentage == 100:
        print("Show Status: Housefull")
    elif occupancy_percentage >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")

    if occupancy_percentage >= 90:
        print("Suggestion: Consider Opening Additional Show")


# Sample Input
total_seats = 200
booked_seats = [1] * 150

seat_occupancy_analyzer(total_seats, booked_seats)
```

```python
#4.
def classify_server_load(cpu_readings):

    total = 0

    # Loop to calculate sum
    for reading in cpu_readings:
        total += reading

    # Average calculation
    average_cpu = total / len(cpu_readings)

    print("Average CPU Load:", int(average_cpu), "%")

    # Classification
    if average_cpu < 50:
        print("Server Status: Normal")
    elif 50 <= average_cpu <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")

cpu_readings = [45, 60, 70, 85, 90]

classify_server_load(cpu_readings)
```

```python
#5.
def classroom_resource_monitor(resource_usage):

    overused_resources = []

    for resource, hours in resource_usage.items():
        if hours > 8:
            overused_resources.append(resource)

    # Display overused resources
    if overused_resources:
        print("Overused Resources:", ", ".join(overused_resources))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")

usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

classroom_resource_monitor(usage)
```

```python
#6.
def event_registration_controller(capacity, total_registrations):

    confirmed = 0
    waitlist = 0

    # Loop to process each registration
    for i in range(total_registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlist += 1

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlist)

    if confirmed == capacity:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")

capacity = 100
registrations = 105

event_registration_controller(capacity, registrations)
```
