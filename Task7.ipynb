```python
#1.
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"

    def turn_on(self):
        self.status = "ON"

    def turn_off(self):
        self.status = "OFF"

    def display_status(self):
        print(f"{self.name} is {self.status}")

# Sample Input
light_name = input("Light Name: ")
action = input("Action: ")

# Create object
light = SmartLight(light_name)

if action.upper() == "ON":
    light.turn_on()
elif action.upper() == "OFF":
    light.turn_off()

# Display Output
light.display_status()
```

```python
#2.
class Employee:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_id_card(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")

# Sample Input
name = input("Employee Name: ")
emp_id = input("Employee ID: ")
department = input("Department: ")

# Create object
employee = Employee(name, emp_id, department)

employee.display_id_card()
```

```python
#3.
class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def display_contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")


# Sample Input
name = input("Contact Name: ")
phone = input("Phone Number: ")

# Create object
contact = Contact(name, phone)

# Display Output
contact.display_contact()
```

```python
#4.
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def print_price_tag(self):
        print(f"Product: {self.name}")
        print(f"Price: ₹{self.price}")

# Sample Input
name = input("Product Name: ")
price = input("Price: ")

# Create object
product = Product(name, price)

# Display Output
product.print_price_tag()
```

```python
#5.
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display_details(self):
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating} / 5")

# Sample Input
name = input("Movie Name: ")
rating = input("Rating: ")

# Create object
movie = Movie(name, rating)

# Display Output
movie.display_details()
```

```python
#6.
class Delivery:
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    def print_delivery_details(self):
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")

# Sample Input
name = input("Customer Name: ")
address = input("Address: ")

# Create object
delivery = Delivery(name, address)

# Display Output
delivery.print_delivery_details()
```
