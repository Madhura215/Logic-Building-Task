```python
#2.
employees={"Ravi":75000,"Anita":68000,"Kiran":72000}
high_name=""
high_salary=0
for name in employees:
  if employees[name]>high_salary:
    high_salary=employees[name]
    high_name=name
print("highest salary:",high_name,"-",high_salary)
```

```python

#3.

# List of numbers
numbers = [45, 22, 89, 10, 66]

# Initialize max and min with first element
maximum = numbers[0]
minimum = numbers[0]

# Loop through list
for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

# Print results
print("List:", numbers)
print("Max:", maximum)
print("Min:", minimum)
```

```python
#4.
# List of product prices
prices = [450, 1200, 899, 1500, 300]
count = 0
for price in prices:
    if price > 1000:
        count += 1
print("Products above 1000:", count)
```

```python
#5.
attendance = ["P", "P", "A", "P", "P"]
present_count = 0
# Count number of "P"
for day in attendance:
    if day == "P":
        present_count += 1
# Calculate percentage
total_days = len(attendance)
percentage = (present_count / total_days) * 100
print("Attendance Percentage:", percentage)
```

```python
#6.
phone_numbers = [9876543210, 9123456789, 9876543210]
# Convert list to set to remove duplicates
unique_numbers = set(phone_numbers)
print("Unique phone numbers:", unique_numbers)
```

```python
#7.
txt = "python"
frequency = {}
for char in txt:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1
print(frequency)
```
