```python
#2.Pass/Fail Analyzer

marks=[45,78,90,33,60]
p_count=0
f_count=0
for mark in marks:
    if mark>=50:
      p_count +=1
    else:
      f_count +=1
print("total students:",len(marks))
print("number of pass students:",p_count)
print("number of fail students:",f_count)
```

```python
#3.Simple Data Cleaner

names=["Alice","bob","CHARLIE"]
clean_names=[]
for name in names:
    clean_name=name.strip().lower()
    clean_names.append(clean_name)
print("clean names:",clean_names)
```

```python

#4.Message Length Analyzer

messages=["Hi","Welcome to the platform","OK"]
for msg in messages:
    length=len(msg)
    print("Message:",msg)
    print("Length:",length)
    if length>10:
      print("status:longer than 10 characters")
    print()
```
