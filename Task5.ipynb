```python
#1.
# Function to analyze post engagement
def analyze_engagement(likes_list):
    total_likes = 0

    # Loop to calculate total likes
    for like in likes_list:
        total_likes += like

    print("Total Likes:", total_likes)

    # Check if post is viral
    if total_likes >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")

likes = [200, 300, 150, 400]   # Total = 1050
analyze_engagement(likes)
```

```python
#3.
# Function to check rainfall adequacy
def rainfall_checker(rainfall_data, required_level):
    total = 0

    # Loop to calculate total rainfall
    for rain in rainfall_data:
        total += rain

    # Calculate average rainfall
    average = total / len(rainfall_data)

    print("Average Rainfall:", int(average))

    if average >= required_level:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")


# Example usage
seasonal_rainfall = [70, 75, 80, 63]
rainfall_checker(seasonal_rainfall, 70)
```

```python
#4.
# Function to detect duplicate usernames
def detect_duplicates(usernames):
    unique_users = set()
    duplicates_found = False

    for user in usernames:
        if user in unique_users:
            duplicates_found = True
            break
        unique_users.add(user)

    if duplicates_found:
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


# Example usage
user_list = ["madhura01", "rahul23", "sneha99", "rahul23"]
detect_duplicates(user_list)
```

```python
#6.
# Function to filter premium crop prices
def premium_crop_filter(prices):
    premium_crops = []

    # Loop to filter prices
    for price in prices:
        if price > 2000:
            premium_crops.append(price)

    print("Premium Crops:", premium_crops)


# Example usage
crop_prices = [1500, 2500, 1800, 3200]
premium_crop_filter(crop_prices)
```

```python
#9.
# Function to check attendance eligibility
def check_attendance(attendance_list):
    total_days = len(attendance_list)
    present_days = 0

    # Loop to count present days
    for status in attendance_list:
        if status == "P":
            present_days += 1

    # Calculate attendance percentage
    percentage = (present_days / total_days) * 100

    print("Attendance Percentage:", percentage)

    # Check eligibility
    if percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")

# Example usage
attendance = ["P", "P", "A", "P", "P"]   # 4 out of 5 = 80%
check_attendance(attendance)
```

```python
#10.
# Function to calculate electricity bill and classify usage
def electricity_bill(units):
    bill = 0

    # Slab calculation
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    # Usage classification
    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"

    return bill, status

# Example usage
units_consumed = 250
total_bill, usage_status = electricity_bill(units_consumed)
print("Units Consumed:", units_consumed)
print("Total Bill: ₹", total_bill)
print("Usage Level:", usage_status)
```
